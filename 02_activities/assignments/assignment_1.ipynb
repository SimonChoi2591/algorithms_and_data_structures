
-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
# Your answer here
def is_valid_brackets(s: str) -> bool:
     
    # use stack to pull each matching pair
    stack_str = []

    # creat those pair dictionary
    brackets_pair = {
        '(':')',
        '[':']',
        '{':'}'
        }
    #print(f"pair test", brackets_pair['('])

    # feed each char 
    for c in s:
        #print(c)
        if c in brackets_pair: # change to use the defined list, more efficient if any change of those pairs in future.
            stack_str.append(c)
            #print(f"stack-", stack_str)

        elif c in brackets_pair.values(): # change to use the defined list.
            #print(len(stack_str))
            #if not len(stack_str):   # move this to the end.
            if not stack_str:
                return False
            t = stack_str.pop()
            #print(f"t-", t, "c-",c)

            if  c != brackets_pair[t]:
                return False 
        
    # adding a final stack emptiness check.  to verify that all open brackets are closed.  
    return len(stack_str) == 0
````


-   Explain why your solution works


````python
# Your answer here
Process the string character by character. 
If the character is an open bracket, push it onto the stack. 
Otherwise, it must be a closing bracket, so check whether it matches the bracket type at the top of the stack. 
If they do not match, the sequence is invalid.
````


-   Explain the problem’s time and space complexity


````python
# Your answer here
The algorithm checks each character once. Each push or pop operation takes constant time, 
so the time complexity is O(n).

In the worst case, all character are opening brackets, so they are stored in the stack. 
The stack grow to the size of the string, so the space complexity is O(n).
````


-   Explain the thinking to an alternative solution (no coding required, but a classmate reading this should be able to code it up based off your text)


````python
# Your answer here
Another approach is search and remove matching bracket pairs repeatedly

Create a copy of the original string (copy_string) so the original input is not modified.
Define the three valid bracket pairs: "()", "[]" and "{}"

Repeatedly scan copy_string to check whether it contains any valid bracket pair.
If a matching pair is found, record its starting index x.
Remove the two characters at positions x and x+1 from copy_string.

Continue this process until no more matching bracket pairs exist, or copy_string becomes empty.

Must include a termination condition to ensure the loop exits when no further changes can be made, 
preventing an infinite loop caused by repeatedly processing the same string.

Return True if copy_string is empty (all brackets were matched and removed).
otherwise, return False, (if any unmatched brackets remain).



The algorithm first creates a copy of the original string, which requires O(n) space. 
During the removal process, an additional temporary space is used to store the modified string, which also requires O(n) space. 
Therefore, the total space complexity is O(n) + O(n), which simplifies to O(n).

Worst case #1: The string contains only opening brackets, with no matching pairs. The algorithm scans the entire copy_string once, taking O(n) time.
Worst case #2: The string contains n/2 opening brackets followed by n/2 closing brackets, where the matching pairs start from the middle of the string. 
Each removal operation requires scanning and creating a modified string of approximately O(n/2). This removal process may repeat n/2 times. 
Therefore, the total time complexity is O(n/2 × n/2), which simplifies to O(n²).

````
